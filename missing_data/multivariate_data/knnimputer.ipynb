## multivariate is using other cols to fill missing values of a col

##

```python
import pandas as pd
```

```python
df = pd.read_csv('../../titanic.csv', usecols=['Age', 'Pclass', 'Survived'])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(df.iloc[:,1:], df.iloc[:,0], test_size=0.2, random_state=42)
```

```python
from sklearn.impute import KNNImputer
imp = KNNImputer(n_neighbors=20, weights='distance')
x_train_trf = imp.fit_transform(x_train)
x_test_trf = imp.fit_transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
lr = LogisticRegression()
lr.fit(x_train_trf, y_train)
y_pred = lr.predict(x_test_trf)
accuracy_score(y_pred, y_test)
```

```python
from sklearn.model_selection import cross_val_score
import numpy as np
np.mean(cross_val_score(lr, x_train_trf, y_train, cv=10))
```
