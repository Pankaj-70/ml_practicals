```python
from sklearn.datasets import make_classification
x, y = make_classification(n_samples = 100, n_features = 2, n_classes = 2, n_informative = 1, n_redundant = 0, class_sep = 10, hypercube = False, n_clusters_per_class = 1, random_state=41)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
plt.scatter(x[:, 0], x[:,1], c = y, cmap = 'winter', s = 100)
plt.show()
```

```python
def step_func(z):
    return 1 if z >= 0 else 0
```

```python
import numpy as np
epochs = 1000

def perceptron(x, y):
    x = np.insert(x, 0, 1, axis = 1)
    weights = np.ones(x.shape[1])
    lr = 0.1                    #learning rate

    for i in range(epochs):
        randomIndex = np.random.randint(0, 100)
        yhat = step_func(np.dot(x[randomIndex], weights))
        weights = weights + lr * (y[randomIndex] - yhat) * x[randomIndex]
    
    return weights[0],weights[1:]
```

```python
intercept_, coeff_ = perceptron(x, y)
```

```python
print(coeff_)
print(intercept_)
```

```python
m = -(coeff_[0] / coeff_[1])
b = -(intercept_ / coeff_[1])
```

```python
x_input = np.linspace(-3, 3, 100)
y_input = m * x_input + b
```

```python
plt.figure(figsize=(10, 6))
plt.plot(x_input, y_input, color = 'red', linewidth = 3)
plt.scatter(x[:, 0], x[:, 1], c = y, cmap = 'winter', s = 100)
plt.ylim(-3, 2.5)
plt.show()
```
