## MICE (Multivariate Imputation By Chained Equations)

```python
import pandas as pd
import requests
from io import StringIO
import numpy as np
```

```python
res = requests.get('https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/refs/heads/main/day40-iterative-imputer/50_Startups.csv').text
df = np.round(pd.read_csv(StringIO(res))[['R&D Spend', 'Administration', 'Marketing Spend', 'Profit']]/10000)
np.random.seed(9)
df_missing = df.drop(columns=['Profit'])
df_missing = df_missing.sample(5)
df_missing.loc[21, 'R&D Spend'] = np.nan
df_missing.loc[37, 'Administration'] = np.nan
df_missing.loc[2, 'Marketing Spend'] = np.nan
df_missing
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.linear_model import LinearRegression
from sklearn.impute import IterativeImputer
imp = IterativeImputer(max_iter=10, estimator=LinearRegression(), random_state=42)
```

```python
pd.DataFrame(imp.fit_transform(df_missing), columns=['R&D Spend', 'Administration', 'Marketing Spend'])
```
