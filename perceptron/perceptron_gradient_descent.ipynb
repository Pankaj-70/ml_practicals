```python
from sklearn.datasets import make_classification
import numpy as np
import matplotlib.pyplot as plt
```

```python
x, y = make_classification(n_samples=100, n_features=2, n_informative=1, n_classes=2, n_clusters_per_class=1, n_redundant=0, hypercube=False, class_sep=15, random_state=41)
y = np.where(y == 0, -1, 1)
```

```python
plt.figure(figsize=(10, 4))
plt.scatter(x[:, 0], x[:, 1], c = y, cmap = 'winter', s = 100)
plt.show()
```

```python
epochs = 1000
def perceptron(x, y):
    w1, w2, b = 1, 1, 1
    lr = 0.1
    for _ in range(epochs):
        for i in range(x.shape[0]):
            z = w1 * x[i][0] + w2 * x[i][1] + b

            if z * y[i] < 0:
                w1 =  w1 + lr * y[i] * x[i][0]
                w2 =  w2 + lr * y[i] * x[i][1]
                b =  b + lr * y[i]


    return w1, w2, b
```

```python
w1, w2, b = perceptron(x, y)
print(w1, w2, b)
```

```python
m = -(w1 / w2)
c = -(b / w2)
print(m, c)
```

```python
x_input = np.linspace(-3, 3, 100)
y_input = m * x_input + c
```

```python
plt.figure(figsize=(10, 6))
plt.plot(x_input, y_input, linewidth=3, color = 'red')
plt.scatter(x[:, 0], x[:, 1], c = y, cmap = 'winter', s = 100)
plt.ylim(-4, 2)
plt.show()
```
