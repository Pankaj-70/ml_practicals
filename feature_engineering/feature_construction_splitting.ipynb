```python
import numpy as np
import pandas as pd
```

## Feature Construction

```python
df = pd.read_csv('../titanic.csv', usecols = ['Age', 'Pclass','SibSp','Parch', 'Survived'])
df.head(5)
```

```python
df['family_size'] = df['Parch'] + df['SibSp'] + 1
df.head()
```

```python
df.dropna(inplace = True)
```

```python
df.drop(columns = ['SibSp', 'Parch'], inplace = True)
df.head()
```

```python
from sklearn.model_selection import train_test_split
x = df.iloc[:,1:]
y = df.iloc[:,0]
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(x_train, y_train)
y_pred = lr.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_pred, y_test)
```

```python
from sklearn.model_selection import cross_val_score
np.mean(cross_val_score(LogisticRegression(), x_train, y_train, scoring='accuracy', cv = 20))
```

```python
def find_family_type(num):
    if num == 1:
        return 0  #single
    elif num <= 3:
        return 1 #small family
    else:
        return 2 #big family
```

```python
df['family_type'] = df['family_size'].apply(find_family_type)
```

```python
df.head()
```

```python
df.drop(columns=['family_size'], inplace = True)
```

```python
x = df.iloc[:,1:]
y = df.iloc[:,0]
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
lr = LogisticRegression()
lr.fit(x_train, y_train)
y_pred = lr.predict(x_test)
```

```python
np.mean(cross_val_score(LogisticRegression(), x_train, y_train, scoring='accuracy', cv = 20))
```
